```python
import json
import boto3
import pandas as pd
from sklearn.model_selection import train_test_split

s3_client = boto3.client('s3')

manifest_key = 'ground-truth-demo/labeled-data/output.manifest'
manifest_obj = s3_client.get_object(Bucket=bucket, Key=manifest_key)
manifest_lines = manifest_obj['Body'].read().decode('utf-8').strip().split('\n')

data = []
for line in manifest_lines:
    obj = json.loads(line)
    image_uri = obj['source-ref']
    label = obj.get('ground_truth_demo', {}).get('annotations')[0]['class_id']
    label_name = obj.get('ground_truth_demo-metadata', {}).get('class-map', {}).get(str(label), 'unknown')
    data.append({'image_uri': image_uri, 'label_id': label, 'label_name': label_name})

df = pd.DataFrame(data)

print(f"Total images: {len(df)}")
print(f"\nClass distribution:")
print(df['label_name'].value_counts())
print(f"\nLabel mapping:")
print(df[['label_id', 'label_name']].drop_duplicates().sort_values('label_id'))

train_df, val_df = train_test_split(df, test_size=0.2, random_state=42, stratify=df['label_id'])

print(f"\nTraining set: {len(train_df)} images")
print(f"Validation set: {len(val_df)} images")
```
